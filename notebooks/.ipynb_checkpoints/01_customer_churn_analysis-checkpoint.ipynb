# Exploratory Data Analysis (EDA)

The objective of EDA is to identify customer churn patterns and uncover business insights from dataset.

## Customer Churn Distribution

This analysis helps to understand the proportion of customers who stayed versus customer who left the company.

```python
df['Churn'].value_counts()
```

```python
df['Churn'].value_counts(normalize=True) *100
```

## Observation

- Majority of customer are retained.
- However, churn percentage is significant and can impact company revenue.
- Customer retention stretigies may help in reducing business loss.

## Churn Visualisation


```python
sns.countplot(x='Churn', data=df)

plt.title('Customer Churn Distribution')
plt.xlabel('Churn')
plt.ylabel('Customer Count')

plt.show()
```

## Visualization Insight

- The number of retained customers is higher than churned customers.
- Despite lower churn counts, customer attrition remains business-critical.

## Contract Type vs Customer Churn

This analysis checks whether contract duration impacts customer retention.


```python
pd.crosstab (df['Contract'],df['Churn'],normalize='index') *100
```

## Observation

- Customers with month-to-month contracts show significantly higher churn rates.
- Logn-term contracts appear to improve customer retention.
- Customer commitment duration may strongly influence churn behavior.


## Customer Tenure Analysis

Customer tenure respresents the duration a customer has stayed with the company.

This analysis helps determine whether long-term customers are less likely to churn.


```python
df.groupby('Churn')['tenure'].mean()
```

## Observation

- Customer with lower tenure are most likely to churn.
- Long-term customers appear more loylal to the company.
- Customer retention strategies may be especially important during early customer stages.

```python
df.groupby('Churn')['tenure'].mean().plot(kind='bar')

plt.title('Average Tenure by Churn')
plt.xlabel('Churn')
plt.ylabel('Average Tenure')

plt.show()
```

## Mothly Charges vs Customer Churn

Thsi analysis evalutes whether higher monthly charges influence customer churn behavior.

```python
df.groupby('Churn')['MonthlyCharges'].mean()
```

## Observation

- Customers with higher monthly charges may show increased churn tendencies.
- Pricing could be a contributing factor affecting customer retention.

```python

df.groupby('Churn')['MonthlyCharges'].mean().plot(kind='bar')

plt.title('Average Monthly Charges by Churn')
plt.xlabel('Churn')
plt.ylabel('MonthlyCharges')

plt.show()
```

## Internet Service vs Customer Churn

This analysis examines whether the type of internet service impacts customer churn behavior.

```python
pd.crosstab(df['InternetService'],df['Churn'],normalize='index') *100
```

## Observation

- Certain internet service categories show higher churn percentages.
- Service quality, pricing, or customer expectations may influence churn behavior.

## Payment Method vs Customer Churn

This analysis evaluates whether payment mentods influence customer retention patterns.

```python
pd.crosstab(df['PaymentMethod'],df['Churn'],normalize='index')*100
```

## Observation

- Some payment methods demonstrate higher churn percentages.
- Automatic payment methods may indicate stronger customer retention behavior.

## Senior Citizen vs Customer Churn

This analysis checks whether senior citizens show different churn behavior compared to non- senior customer.

```python
pd.crosstab(df['SeniorCitizen'],df['Churn'],normalize='index').mul(100).round(2)
```

## Observation

- Senior citizens may display different churn tendencies compared to younger customers.
- Customer demographics can influence retention behavior.

# Correlation Analysis

COrrelatin analysis helps identify relationship between numerical veriables in the dataset.



```python
df['Churn_Numeric'] = df['Churn'].map({
    'Yes': 1,
    'No': 0
})
```

```python
df[['tenure', 'MonthlyCharges', 'SeniorCitizen', 'Churn_Numeric']].corr()
```

## Correlation Insights

- Customer tenure shows a negative relationship with churn.
- Customers staying longer are less likely to leave the company.
- Monthly charges may have a mild positive relationship with churn.

```python
plt.style.use('ggplot')
```

## Customer Tenure Distribution

```python
plt.figure(figsize=(8,5))

plt.hist(df['tenure'], bins=30)

plt.title('Customer Tenure Distribution')
plt.xlabel('Tenure')
plt.ylabel('Customer Count')

plt.show()
```

## Observation

- Customer tenure distribution shows a mix of short-term and long-term customers.
- A noticeable number of customers appear within lower tenure ranges.

## Correlation Heatmap

```python
corr = df[['tenure',
           'MonthlyCharges',
           'SeniorCitizen',
           'Churn_Numeric']].corr()

plt.figure(figsize=(8,5))

sns.heatmap(corr, annot=True)

plt.title('Correlation Heatmap')

plt.show()
```

## Observation

- Customer tenure demonstrates a negative relationship with churn.
- Monthly charges show a weaker positive relationship with churn.
- Correlation values suggest tenure is an important retention indicator.

# Key Business Insights

1. Customers with month-to-month contracts demonstrate significantly higher churn rates.
2. Customers with lower tenure are more likely to leave the company.
3. Higher monthly charges may contribute to increased churn behavior.
4. Certain payment methods and internet services show elevated churn percentages.
5. Long-term customer relationships strongly improve retention.

# Business Recommendations

- Encourage customers to shift toward long-term contracts.
- Introduce retention campaigns for new customers.
- Offer loyalty discounts for high-risk churn segments.
- Improve service quality for customer groups with elevated churn.
- Monitor pricing strategies to reduce customer dissatisfaction.
